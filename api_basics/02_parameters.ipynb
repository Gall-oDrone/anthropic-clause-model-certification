### System Prompt

A system prompt is a way to provide context, instructions, and guidelines to Caluse before presenting it with a question or task. By using a system prompt, you can set the stage for the conversation, specifying Claude's role, personality, tone, or any other relevant information that will help it better understand and respond to the user's input

```python
import boto3
from consts_api import *
from utils_api import get_model_response
```

```python
bedrock_client = boto3.client(service_name='bedrock-runtime', region_name="us-west-2")

# Send the message.
response = bedrock_client.converse(
    modelId=MODEL_ID,
    system=[{"text": "You are a very serious assistant who does not tolerate joking."}],
    messages=[{
        "role": "user",
        "content": [{"text": "tell me a joke"}]
    }]
)
```

### Max Tokens

The max_tokens paramters allows us to set an upper limit on how many tokens Claude generates for us. As an illustration, suppose we ask Claude to write us a opem and set max_tokens to 10. Claude will start generating tokens for us and immediately stop as soon as it hits 10 tokens. This will often lead to truncated or incomplete outputs. Let's try it!

```python
bedrock_client = boto3.client(service_name='bedrock-runtime', region_name="us-west-2")

inference_config={"maxTokens":30}

# Send the message.
response = bedrock_client.converse(
    modelId=MODEL_ID,
    messages=[{
        "role": "user",
        "content": [{"text": "tell me a joke"}]
    }]
    inference_config=inference_config
)
```

### Stop Sequences

Another important parameter we haven't seen yet is stop_sequence which allows us to provide the model with a set of strings that, when encountered in the generated response, cause the generation to stop. They are essentially a way of telling Claude, "if you generate this sequence, stop generating anything else!"

```python
messages = [{
    "role": "user",
    "content": [{"text": "Write code to generate prime numbers in Python. Put your code inside of <python> tags."}]
}]

response = bedrock_client.converse(
    modelId=MODEL_ID,
    system=[{"text":"generate a poem"}],
    messages=messages,
)
```

```python
get_model_response(response)
```

```python
messages = [{
    "role": "user",
    "content": [{"text": "Write code to generate prime numbers in Python. Put your code inside of <python> tags."}]
},
{
    "role": "assistant",
    "content": [{"text":"<python>"}]
}]

response = bedrock_client.converse(
    modelId=MODEL_ID,
    system=[{"text":"generate a poem"}],
    messages=messages,
    inferenceConfig={"stopSequences":["</python>"]}
)
```

```python
get_model_response(response)
```

### Temperature

The temperature parameter is used to control the "randomness" and "creativity" of the generated responses. It ranges from 0 to 1, with higher values resulting in more diverse and unpredictable responses with variation in phrasing. Lower temperatures can result in more deterministic outputs that stick to the most probable phrasing and answers.

```python
messages = [{
    "role": "user",
    "content": [{"text": "Come up with a name for an alien planet. Respond with a single word."}]
}]

def demonstrate_temperature():
    temperatures = [0, 1]
    for temp in temperatures:
        print(f"Prompting Claude five times with temperature of {temp}")
        print("==================")
        for i in range(5):
            inference_config={"temperature":temperature}

            # Send the message.
            response = bedrock_client.converse(
                modelId=MODEL_ID,
                messages=messages,
                inferenceConfig=inferenceConfig,
            )
            print(f"Response {i+1}: {get_model_response(response)}")
```
