## Multiple Images

````python
import mimetypes

def create_image_message(image_path):
    # Open the image file in "read binary" mode
    with open(image_path, "rb") as image_file:
        # Read the contents of the image as a bytes object
        binary_data = image_file.read()
    
    # Get the MIME type of the image  based on its file extension
    mime_type, _ = mimetypes.guess_type(image_path)

    sub_type = mime_type.split("/")[-1]

    # Create the image block
    image_block = {
        "image": {
            "format": sub_type,
            "source": {
                "bytes": binary_data
            }
        }
    }

    return image_block
````

````python
message = {
    "role": "user",
    "content": [
        {"text": "Identify the subject of each of theses images, using a single word for each image."},
        create_image_message("./images/animal1.png"),
        create_image_message("./images/animal2.png"),
        create_image_message("./images/animal3.png"),
    ]
}

bedrock_client = boto3.client(service_name="bedrock-runtime",region_name="us-west-2")

# Send the message
response = bedrock_client.conerse(
    modelId=MODEL_ID
    messages=[message],
)

print(response)
````
